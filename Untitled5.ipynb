```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# База данных о странах и их природе
NATURE_DATA = {
    "Горы и Вулканы": {
        "country": "Исландия 🇮🇸",
        "description": "Страна льда и огня. Здесь невероятные водопады, вулканы, гейзеры и ледники.",
        "places": ["Водопад Скогафосс", "Гора Киркьюфетль", "Черный пляж Рейнисфьяра", "Ледниковая лагуна"]
    },
    "Леса и Джунгли": {
        "country": "Коста-Рика 🇨🇷",
        "description": "Мировой лидер по биоразнообразию. Вечнозеленые дождевые и облачные леса.",
        "places": ["Заповедник Монтеверде", "Национальный парк Тортугеро", "Вулкан Ареналь"]
    },
    "Фьорды и Озера": {
        "country": "Норвегия 🇳🇴",
        "description": "Главное место в мире для любителей суровых северных пейзажей, фьордов и гор.",
        "places": ["Гейрангер-фьорд", "Скала Язык Тролля (Trolltunga)", "Лофотенские острова"]
    },
    "Океан и Пляжи": {
        "country": "Индонезия (Бали, Комодо) 🇮🇩",
        "description": "Потрясающие рифы, океанские пейзажи, рисовые террасы и тропические острова.",
        "places": ["Остров Комодо", "Нуса Пенида", "Водопад Секумпул"]
    },
    "Уникальный Животный Мир": {
        "country": "Новая Зеландия 🇳🇿",
        "description": "Разнообразие ландшафтов: от альпийских лугов до геотермальных источников и фьордов.",
        "places": ["Милфорд-Саунд", "Национальный парк Тонгариро", "Озеро Текапо"]
    }
}

# Виджет выпадающего списка
dropdown = widgets.Dropdown(
    options=list(NATURE_DATA.keys()),
    description='Природа:',
    style={'description_width': 'initial'}
)

output = widgets.Output()

def on_change(change):
    if change['type'] == 'change' and change['name'] == 'value':
        category = change['new']
        data = NATURE_DATA[category]

        with output:
            clear_output(wait=True)
            places_html = "".join([f"<li>{place}</li>" for place in data['places']])
            display(HTML(f"""
                <div style="border: 2px solid #2FA572; border-radius: 10px; padding: 15px; margin-top: 10px; max-width: 450px; font-family: sans-serif;">
                    <h2 style="color: #2FA572; margin-top: 0;">{data['country']}</h2>
                    <p><b>Описание:</b> {data['description']}</p>
                    <p><b>Главные места:</b></p>
                    <ul>{places_html}</ul>
                </div>
            """))

# Привязываем событие выбора
dropdown.observe(on_change)

# Отображаем элементы
display(dropdown)
display(output)

# Показываем первое значение при запуске
on_change({'type': 'change', 'name': 'value', 'new': dropdown.value})
```
